```python
import pickle
import numpy as np
```

```python
with open('model.bin', 'rb') as f_in:
    pipeline = pickle.load(f_in)
```

```python
print("Pipeline type:", type(pipeline))
```

```python

# Check individual components
if hasattr(pipeline, 'steps'):
    print("\nPipeline steps:")
    for name, step in pipeline.steps:
        print(f"  {name}: {type(step)}")
        if hasattr(step, '__sklearn_version__'):
            print(f"    sklearn version: {step.__sklearn_version__}")
```

```python

# Try to get more info
print("\nPipeline attributes:")
for attr in dir(pipeline):
    if 'version' in attr.lower() or 'sklearn' in attr.lower():
        print(f"  {attr}: {getattr(pipeline, attr, 'N/A')}")
```

```python
def predict_single(re_property):
    result = pipeline.predict(re_property)
    
    restored_value = np.exp(result) - 1
    return round(float(restored_value.item()),2)
```
