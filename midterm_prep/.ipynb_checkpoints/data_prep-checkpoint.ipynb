<details>
<summary><strong>🏠 Property Identification</strong></summary>

- **PID**: Unique 10‐digit parcel number. First 2 digits are the ward, digits 3 to 7 are the parcel, and digits 8 to 10 are the sub‐parcel
- **CM_ID**: 10‐digit parcel number of Condo Main, which houses all related condo units
- **GIS_ID**: Primary GIS ID
- **BLDG_SEQ**: Building sequence of parcel
- **NUM_BLDGS**: Total number of buildings of parcel

</details>

<details>
<summary><strong>📍 Location & Address</strong></summary>

- **ST_NUM**: Street number of parcel
- **ST_NAME**: Street name of parcel
- **UNIT_NUM**: Specific unit number within a housing complex
- **CITY**: City or Town of parcel
- **ZIP_CODE**: Zip code of parcel

</details>

<details>
<summary><strong>🏘️ Property Classification</strong></summary>

- **LUC**: Land Use Code

    - **101**: SINGLE FAM DWELLING
    - **102**: RESIDENTIAL CONDO
    - **103**: MOBILE HOME
    - **104**: TWO-FAM DWELLING
    - **105**: THREE-FAM DWELLING
    - **106**: ADD'L RES IMPROVEMENT
    - **107**: OTHER RESIDENTIAL
    - **108**: CONDO PARKING
    - **109**: MULTIPLE BUILDINGS
    - **110**: CONDO STORAGE

- **LU**: Type of Property (land use)
  - **A**: Residential 7 or more units
  - **AH**: Agricultural/Horticultural
  - **C**: Commercial
  - **CC**: Commercial condominium
  - **CD**: Residential condominium unit
  - **CL**: Commercial land
  - **CM**: Condominium main (physical structure housing all related condo units with no assessed value)
  - **CP**: Condo parking
  - **E**: Tax‐exempt
  - **EA**: Tax‐exempt (121A)
  - **I**: Industrial

## Getting the Data

```python
import pandas as pd
import warnings
warnings.filterwarnings("ignore")
df = pd.read_csv('fy2025-property-assessment-data_12_30_2024.csv', dtype={'ZIP_CODE': str})
```

```python
rows, cols = df.shape
print(f"The dataset contains {rows:,} records with {cols} columns")
```

### Column Analysis

```python
# Clean the columns
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')
```

```python
df.columns
```

### Renaming Columns

```python
# Column renaming dictionary
column_rename_map = {
    'pid': 'parcel_id',
    'cm_id': 'condominium_id', 
    'gis_id': 'geographic_id',
    'st_num': 'street_num',
    'st_num2': 'street_num2',
    'st_name': 'street_name',
    'bldg_seq': 'bldg_sequence',
    'lu': 'land_use_type',
    'luc': 'land_use_code',
    'lu_desc': 'land_use_description',
    'own_occ': 'owner_occupied',
    'res_floor': 'residential_floor',
    'cd_floor': 'condo_floor', 
    'res_units': 'residential_units',
    'com_units': 'commercial_units',
    'rc_units': 'res_or_commercial_units',
    'hlf_bth': 'half_bth',
    'tt_rms': 'total_num_rooms',
    'prop_view': 'property_view'
}

# Apply the renaming
df = df.rename(columns=column_rename_map)
```

```python
df.head(4)
```

```python
df.columns
```

### Filtering for Residential Units Only

```python
# focus on residential properties only. 
residential_list = ['CD', 'R1', 'R2', 'R3', 'R4', 'RC']
```

- A: Residential 7 or more units
- CD: Residential condominium unit
- R1: Residential 1‐family
- R2: Residential 2‐family
- R3: Residential 3‐family
- R4: Residential 4 or more family
- RC: Mixed use (residential and commercial)

```python
df_residential = df[df['land_use_type'].isin(residential_list)]
```

### Getting Rid of Irrelevant Columns 

- Columns to be removed are the ones that pertain to the mailing address of the owner (fields that start with `_mail`)
- Bathroom Style (`bhthrm_style`) and kitchen style (`kitchen_style`) columns are to be removed since overall condition (`overall_cond`) field is available
- Bedroom Condition (`bdrm_cond`), Interior Condition of parcel‐Residential (`int_cond`), Exterior condition of parcel (`ext_cond`), and Interior Wall Condition‐Residential (`int_wall`) columns will also be removed since overall condition (`overall_cond`) field is available and will allow for a simpler modeling
- Total assessed land value (`land_value`) column is to be removed because we are dealing with residential properties and not land
- Number of Residential/Commercial Units in a condominium building (`res_or_commercial_units`)  can either increase value (convenience, rental income potential) or decrease it (noise, traffic, commercial activity), making the signal unclear. Also, the analysis targets individual residential property values - what matters is the specific unit's characteristics, not the building's overall commercial mix
- Number of commercial units in a condominium building (`commercial_units`) since the analysis targets residential property buyers and investors, the number of commercial units in a building doesn't impact residential unit valuation decisions
- Number of residential units in a condominium building (`residential_units`) since the assumption is that one unit is examed at a time
- Columns that contain information about condominium id (`condominium_id`), parcel id (`parcel_id`) and Geographic information system id (`geographic_id`) since the information about zip code (`zip_code`) is available
- Anything about street number(s) (`street_num` and `street_num2`), street name (`street_name`), unit number (`unit_num`), and city (`city`) since the information about zip code (`zip_code`) is available
- `sfyi_value` - this field has not been described at all. If the field is not described, it would not be wise to use it
- Structure Class (`structure_class`) not only has 131882 fields missing but also structure class is highly correlated with year built - older buildings are typically wood frame, newer ones use steel/concrete. Also, The construction material itself doesn't directly determine value - what matters is the resulting condition, durability, and maintenance costs (captured by condition fields)
- Building Sequence (`bldg_sequence`) only applies to multi-building parcels, making it irrelevant for most residential properties and the sequence number itself doesn't influence what buyers are willing to pay
- Total number of buildings of parcel (`num_bldgs`) is excluded as well since most residential parcels have only one building, creating minimal variation
- Total assessed building value (`bldg_value`) is not relevant since the the target value of `total_value` has been selected
- Parcel's land area in square feet (legal area) (`land_sf`) and living area square footage of the property (`living_area`) are not relevant since the field Gross floor area (`gross_area`) is available and should be used for modeling. It is also the field that is actually used for the tax purposes.

- Exterior Siding Material (`ext_fnished`) is not relevant because siding type affects maintenance costs but has minimal impact on property value in Boston's mature market. Also, in a historic city like Boston, architectural character matters more than specific siding materials.
- Corner unit (`corner_unit`) is a limited-scope field being applicable only to condominiums, excluding single-family homes and other property types in the scope of the analysis
- Property view (`property_view`) is not relevant because the property view is subjective to the property assessor's view. People who are going to purchase the property might have a completely different view. 
- Roof Cover Material (`roof_cover`) and Roof Structure Types (`roof_structure`) are also not very applicable in the case of purchase. The only feature that would impact the purchase of the property is how recently the roof was renovated. This field is not available in the dataset. 
- The location a condominium unit in the building is facing (`orientation`) should also be excluded because this field only applies to condominium units, excluding single-family homes, two-family homes, and other residential property types that make up a significant portion of our dataset
- Heating System (individually controlled, common or self‐contained) (`heat_system`) has 66081 and there is already information available about heat with far less missing data - Heating type (`heat_type`)
- Kitchen Type ‐ Residential (`kitchen_type`) since there is a numeric representation of kitchens, which is under the field `kitchen`

```python
unwanted_columns = (
    [col for col in df_residential.columns if col.startswith('mail_')] + 
    [col for col in df_residential.columns if col.startswith('bthrm_style')] +
    [col for col in df_residential.columns if col.startswith('kitchen_style')] +
    ['bdrm_cond', 'int_cond', 'ext_cond', 'int_wall',  'land_value', 'res_or_commercial_units', 'commercial_units', 'residential_units']+
    ['condominium_id', 'parcel_id', 'geographic_id', 'street_num', 'street_num2', 'street_name', 'unit_num', 'city'] +
    ['sfyi_value', 'structure_class', 'bldg_sequence', 'num_bldgs', 'bldg_value', 'land_sf', 'living_area'] +
    ['ext_fnished', 'corner_unit', 'property_view', 'roof_cover' , 'roof_structure', 'orientation', 'heat_system', 'kitchen_type']
     
)

df_dropped_columns = df_residential.drop(columns=unwanted_columns)
```

```python
df_dropped_columns.head(3)
```

### Filtering Mixed use (residential and commercial) Properties (RC) to Only Those with Room Data

```python
df_dropped_columns_filtered = df_dropped_columns[~((df_dropped_columns['land_use_type'] == 'RC') & (df_dropped_columns['total_num_rooms'].isna()))]
```
