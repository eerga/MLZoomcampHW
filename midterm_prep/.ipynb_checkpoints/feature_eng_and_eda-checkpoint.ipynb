```python
numerical_columns = [
    'land_use_code',       # Land use code (numeric version)
    'residential_floor',   # Number of residential floors
    'condo_floor',         # Condo floor number
    'gross_area',          # Gross area in sq ft
    'total_value',         # Property value (target variable)
    'gross_tax',           # Tax amount (needs cleaning - has $ and commas)
    'yr_built',            # Year built
    'yr_remodel',          # Year remodeled
    'bed_rms',             # Number of bedrooms
    'full_bth',            # Number of full bathrooms
    'half_bth',            # Number of half bathrooms
    'total_num_rooms',     # Total number of rooms
    'kitchens',            # Number of kitchens
    'fireplaces',          # Number of fireplaces
    'num_parking'          # Number of parking spaces
]

categorical_columns = [
    'zip_code',             # Zip code of parcel
    'land_use_type',        # Type of Property (land use)
    'land_use_description', # Land Use Description
    'bldg_type',            # Building Type and Style
    'owner_occupied',       # "Y" = owner receives residential exemption as an owner‐occupied property
    'overall_cond',         # Overall condition of parcel
    'heat_system',          # Heating System (individually controlled, common or self‐contained)
    'ac_type'               # Air Conditioning Type‐Residential
]
```

```python
df['owner_occupied'] = df['owner_occupied'].map({'N': 0, 'Y': 1})
```

```python
# Fill missing year_remodel with yr_built values
df['yr_remodel'] = df['yr_remodel'].fillna(df['yr_built'])
```

```python
# Fill in missing values with 0.0 for  halfbath, fireplaces, and number of parking spaces
# Since for Boston this is a reasonable assumption
columns_to_fill = ['half_bth', 'fireplaces', 'num_parking']

# Method 1: Fill specific columns with 0.0
df[columns_to_fill] = df[columns_to_fill].fillna(0.0)
```

```python
# Ensuring total_value and gross_tax are numeric values
df['total_value'] = (df['total_value']
                                .str.strip()           # Remove leading/trailing spaces
                                .str.replace(',', '')   # Remove commas
                                .str.replace('$', '')   # Remove dollar signs
                                .astype(float))

df['gross_tax'] = (df['gross_tax']
                                .str.strip()           # Remove leading/trailing spaces
                                .str.replace(',', '')   # Remove commas
                                .str.replace('$', '')   # Remove dollar signs
                                .astype(float))
```

```python
missing_values = df.isna().sum()
columns_with_missing = missing_values[missing_values > 0]; columns_with_missing
```

## Feature Engineering Section

### Floors

```python
# logic here is that typically the max number of the two is the actual floor number of the residence
df['floor_number'] = df[['residential_floor', 'condo_floor']].max(axis=1)

columns_to_drop = (
    ['residential_floor', 'condo_floor']
)

df = df.drop(columns=columns_to_drop)
df = df.dropna(subset=['floor_number'])
```

```python
missing_values = df.isna().sum()
columns_with_missing = missing_values[missing_values > 0]; columns_with_missing
```

### Clean Row Values

```python
categorical_columns = list(df.dtypes[df.dtypes == 'object'].index)

for c in categorical_columns:
    df[c] = (df[c]
             .str.strip()                           # Remove leading/trailing spaces
             .str.lower()                           # Convert to lowercase
             .str.replace(r'-', '_', regex=True)    # Replace hyphens with underscores
             .str.replace(r'\s+', '_', regex=True)  # Replace spaces with underscores
             .str.replace(r'_{2,}', '_', regex=True) # Replace multiple underscores with single
             .str.replace(r'^_|_$', '', regex=True) # Remove leading/trailing underscores
            )
```

### Examine zipcode by median total value and total count of inventory

```python
# Create zip code statistics
zip_code_stats = df_filtered.groupby('zip_code').agg({
    'total_value': ['median', 'count']
}).round(0)

zip_code_stats.columns = ['median_price', 'inventory_count']

# Create price and inventory percentiles for classification
zip_code_stats['price_percentile'] = zip_code_stats['median_price'].rank(pct=True)
zip_code_stats['inventory_percentile'] = zip_code_stats['inventory_count'].rank(pct=True)
```

```python
def classify_zipcode_4_tiers(row):
    """Classify zip codes into 4 tiers based on price and inventory"""
    price_pct = row['price_percentile']
    inv_pct = row['inventory_percentile']
    
    # Premium: Very high price (top 25%) OR Low inventory + High price
    if price_pct >= 0.75 or (inv_pct <= 0.3 and price_pct >= 0.6):
        return 'premium'
    
    # High-End: High price (50th-75th percentile) 
    elif price_pct >= 0.5 and price_pct < 0.75:
        return 'high_end'
    
    # Budget: Low price (bottom 40%) AND High inventory (top 50%)
    elif price_pct <= 0.4 and inv_pct >= 0.5:
        return 'budget'
    
    # Mid-Market: Everything else
    else:
        return 'mid_market'

# Apply the 4-tier classification
zip_code_stats['tier_4_category'] = zip_code_stats.apply(classify_zipcode_4_tiers, axis=1)

# Sort by median price
zip_code_stats_sorted = zip_code_stats.sort_values('median_price')

print("📊 4-TIER ZIP CODE CLASSIFICATION")
print("="*70)
print(zip_code_stats_sorted[['median_price', 'inventory_count', 'tier_4_category']])

# Create the dictionary you requested
avg_value_by_zipcode_4tier = zip_code_stats_sorted['median_price'].to_dict()
print(f"\n📋 Median value by zip code (sorted):")
print(avg_value_by_zipcode_4tier)

# Apply to main dataframe
zip_to_tier_4 = dict(zip(zip_code_stats.index, zip_code_stats['tier_4_category']))
df_filtered['zipcode_4tier'] = df_filtered['zip_code'].map(zip_to_tier_4)

# Summary of the 4-tier classification
print(f"\n🏠 4-TIER SUMMARY:")
print("="*50)

tier_summary = df_filtered.groupby('zipcode_4tier').agg({
    'total_value': ['count', 'median', 'mean', 'std']
}).round(0)

for tier in ['budget', 'mid_market', 'high_end', 'premium']:
    if tier in tier_summary.index:
        count = tier_summary.loc[tier, ('total_value', 'count')]
        median_val = tier_summary.loc[tier, ('total_value', 'median')]
        mean_val = tier_summary.loc[tier, ('total_value', 'mean')]
        std_val = tier_summary.loc[tier, ('total_value', 'std')]
        
        # Get zip codes in this tier
        tier_zips = zip_code_stats[zip_code_stats['tier_4_category'] == tier].index.tolist()
        avg_inventory = zip_code_stats[zip_code_stats['tier_4_category'] == tier]['inventory_count'].mean()
        
        print(f"\n💰 {tier.upper().replace('_', ' ')}:")
        print(f"   Properties: {count:,}")
        print(f"   Median Value: ${median_val:,.0f}")
        print(f"   Mean Value: ${mean_val:,.0f} (±${std_val:,.0f})")
        print(f"   Avg Inventory: {avg_inventory:.1f} properties/zip")
        print(f"   Zip Codes: {', '.join(map(str, tier_zips))}")

# Visualization of 4-tier classification
fig, axes = plt.subplots(1, 2, figsize=(18, 6))

# Colors for 4 tiers
tier_4_colors = {
    'budget': 'green', 
    'mid_market': 'blue', 
    'high_end': 'orange', 
    'premium': 'red'
}

# Scatter plot
for tier in zip_code_stats['tier_4_category'].unique():
    tier_data = zip_code_stats[zip_code_stats['tier_4_category'] == tier]
    axes[0].scatter(tier_data['inventory_count'], tier_data['median_price'], 
                   label=tier.replace('_', ' ').title(), 
                   alpha=0.7, c=tier_4_colors.get(tier, 'gray'), s=100)
    
    # Add zip code labels
    for zip_code, row in tier_data.iterrows():
        axes[0].annotate(str(zip_code), 
                        xy=(row['inventory_count'], row['median_price']),
                        xytext=(5, 5), textcoords='offset points',
                        fontsize=9, alpha=0.8,
                        bbox=dict(boxstyle='round,pad=0.2', 
                                 fc=tier_4_colors.get(tier, 'gray'), alpha=0.3))

axes[0].set_xlabel('Inventory Count', fontsize=12, fontweight='bold')
axes[0].set_ylabel('Median Price ($)', fontsize=12, fontweight='bold')
axes[0].set_title('4-Tier Zip Code Classification', fontsize=14, fontweight='bold')
axes[0].legend(loc='lower right')
axes[0].grid(True, alpha=0.3)
axes[0].yaxis.set_major_formatter(plt.FuncFormatter(lambda x, p: f'${x:,.0f}'))

# Box plot
sns.boxplot(data=df_filtered, x='zipcode_4tier', y='total_value', 
           hue='zipcode_4tier', palette=tier_4_colors, legend=False, ax=axes[1])
axes[1].set_xlabel('Tier', fontsize=12, fontweight='bold')
axes[1].set_ylabel('Property Value ($)', fontsize=12, fontweight='bold')
axes[1].set_title('Property Value Distribution by 4-Tier Classification', fontsize=14, fontweight='bold')
axes[1].yaxis.set_major_formatter(plt.FuncFormatter(lambda x, p: f'${x:,.0f}'))
axes[1].grid(True, alpha=0.3)
axes[1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()

print(f"\n✅ Created 4-tier classification and applied to df_filtered['zipcode_4tier']")

# Distribution check
print(f"\n📊 4-TIER DISTRIBUTION:")
print(df_filtered['zipcode_4tier'].value_counts().sort_index())
```

### Fixing the tax field

In Boston, people qualify for lower taxes when an owner occupies the property. The adjustment value as of 2025 is 3,984.21. [Information source](https://www.boston.gov/departments/assessing/residential-exemption)

```python
# Create the adjusted_tax field
df_filtered['adjusted_tax'] = df_filtered['gross_tax'].copy()

# For owner-occupied properties (owner_occupied == 1), subtract $3,984.21
df_filtered.loc[df_filtered['owner_occupied'] == 1, 'adjusted_tax'] = (
    df_filtered.loc[df_filtered['owner_occupied'] == 1, 'gross_tax'] - 3984.21
)
```

```python
df_filtered = df_filtered.drop(columns=['gross_tax'])
```
